```python
print(5 & 3)
```

 #####  5=0101
 #####  3=0011
 #####  1=0001 (using (&) bitwise operator)


```python
print(5 | 3)
```

 #####  5=0101
 #####  3=0011
 #####  7=0111 (using ( | ) bitwise operator)


```python
print(5 ^ 3)
```

 #####  5=0101
 #####  3=0011
 #####  6=0110 (using (^) bitwise operator)


```python
print(~5) 
```

##### 5= 0000 0101
##### ~5= 1111 1010 (MSB=1 hence the ans is in negative)
##### 1's com= 0000 0101
##### 2's com= 0000 0110 (+1) final ans=-6 (MSB=1) 

```python
print(5 << 2)
```

#####   by formula:
#####           num * 2^value
#####              5 * 2^2=20

```python
print(5 >> 2)
```

#####   by formula:
#####           num // 2^value
#####              5 // 2^2=1
